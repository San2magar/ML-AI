## 1. Are there any missig values in key columns. If so,how should they be handled ?

```python
missing_columns = []
for col in df.columns:
    if	df[col].isnull().sum()>0:
        missing_columns.append(col)
        
print("Missing	Columns	are",	missing_columns)
for	col	in	missing_columns:
    mode_value	=	df[col].mode()[0]
    df[col]	=	df[col].fillna(mode_value)
    
df.isnull().sum()
```

## Data	Visualization	Questions

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df.columns = df.columns.str.replace('\xa0', ' ', regex=False).str.strip()
```

### 1.	What	are	the	top	5	highest	grossing	tours(actual	gross	and	adjusted	gross)?(Bar	chart)

```python
#Actual gross
df.sort_values('Actual gross', ascending=False).head(5).plot(
    kind='bar', x='Tour title', y='Actual gross', 
    title='Top 5 Highest Grossing Tours (Actual Gross)', 
    legend=False
)
plt.ylabel('Actual Gross($)')
plt.show()
```

```python
# Adjusted gross
df.columns = df.columns.str.replace('\xa0', ' ', regex=False).str.strip()
```

```python
df.sort_values('Adjusted gross (in 2022 dollars)', ascending=False).head(5).plot(
    kind='bar',	x='Tour title',	y='Adjusted gross (in 2022 dollars)', 
    title='Top 5 Highest Grossing Tours (Adjusted Gross (in 2022 dollars))', 
    legend=False
)
plt.ylabel('Adjusted Gross (in 2022 dollars)($)')
plt.show()
```

### 12.	How	does	the	number	of	shows	correlate	with	total	gross	and	average	gross?(Scatter	plot)

```python
plt.scatter(df['Shows'], df['Actual gross'], color='r')
plt.title('Number of Shows vs Actual Gross')
plt.xlabel('Number of Shows')
plt.ylabel('Actual Gross ($)')
plt.show()

plt.scatter(df['Shows'], df['Average gross'], color='g')
plt.title('Number of Shows vs Average Gross')
plt.xlabel('Number of Shows')
plt.ylabel('Average Gross per Show ($)')
plt.show()
```

### 13.	Which	artist	appear	most	frequently	in	top	20	highest	grossing	tours?(Bar	chart	or	count	plot)

```python
top20 = df.sort_values('Actual gross', ascending=False).head(20)

plt.figure(figsize=(12,6))
sns.boxplot(data=top20, x='Artist', y='Actual gross')
plt.title('Actual Gross Distribution of Top 20 Highest Grossing Tours by Artist')
plt.ylabel('Actual Gross ($)')
plt.show()
```
