## Line plot basics

```python
import numpy as np
from matplotlib import pyplot as plt 
import matplotlib.pyplot as plt
```

```python
x=np.arange(1,11)
print(x)
```

```python
y=2*x**2
```

```python
print(y)
```

```python
plt.plot(x,y)
plt.show()
```

## adding titles and lables

```python
plt.plot(x,y)
plt.title("Line Plot of x and y variable")
plt.xlabel("x-label")
plt.ylabel("y-label")
plt.show()
```

## changing line aesthetics

```python
plt.plot(x,y,color="orange",linestyle="--",linewidth=3)
plt.title("Line Plot of x and y variable")
plt.xlabel("x-label")
plt.ylabel("y-label")
plt.show()
```

## adding two lines in the same plot

```python
y1=2*x
y2=3*x
```

```python
plt.plot(x,y1,color="g",linewidth=3)
plt.plot(x,y2,color="blue",linestyle=":",linewidth=5)

plt.grid(True)
```

## adding subplots horiontally

```python
plt.subplot(1,2,1)
plt.plot(x,y1,color="g",linewidth=3)

plt.subplot(1,2,2)
plt.plot(x,y1,color="r",linewidth=3)

plt.show()
```

## adding subplots vertically

```python
plt.subplot(2,1,1)
plt.plot(x,y1,color="g",linewidth=3)

plt.subplot(2,1,2)
plt.plot(x,y1,color="r",linewidth=3)

plt.show()
```

```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv(r'D:\Data_science\Day_4\Data_sets/number.csv')
```

```python
## Apply feature engineering to create 'p' based on the condition
def compute_p(row):
    if row['x']>1 and row['y']>1 and row['z']>1:
        return 3 * row['x'] + 2 * np.exp(row['y']) + 5 * row['z']
    else:
        return 0  #if the condition isnot met

df['p'] = df.apply(compute_p,axis = 1)
```

```python
df.head(10)
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
#plot p vs x
plt.figure(figsize=(10,6))
plt.plot(df['x'],df['p'], marker='o', linestyle="-",color="b")
plt.title('Line plot if p vs x')
plt.xlabel("x")
plt.ylabel('p')
plt.grid(True)
plt.show()

#plot p vs y
plt.figure(figsize=(10,6))
plt.plot(df['y'],df['p'], marker='*', linestyle="-",color="r")
plt.title('Line plot if p vs y')
plt.xlabel("y")
plt.ylabel('p')
plt.grid(True)
plt.show()

# plot p vs z
plt.figure(figsize=(10,6))
plt.plot(df['z'],df['p'], marker='h', linestyle="-",color="g")
plt.title('Line plot if p vs z')
plt.xlabel("z")
plt.ylabel('p')
plt.grid(True)
plt.show()
```

## Bar Plot

```python
student={"Aayush Raj Regmi":55, "Manjil Bishwakarma":69,"Riwaz Rayamajhi":30,"Prince Chy":29,"Santu Jhankri Magar":39,"Sambad Khatiwada":35,
         "Prasanna Shakya":26, "Sajal Maharjan":34, "Mango Ghimire":41, "Sexy Laidey":50}
print(student)
```

```python
names= list(student.keys())
values=list(student.values())
```

```python
plt.bar(names,values)
plt.figure(figsize=(300,250))
plt.show()
```

## adding titles and lables to the bar plot

## Horizontal barplot

```python
plt.barh(names,values, color="purple")
plt.title("Bar Plot")
plt.xlabel("Name of a student")
plt.ylabel("marks of a student")
plt.grid(True)
plt.show()
```

## Creating a Basic Scatter Plot

```python
import matplotlib.pyplot as plt
```

```python
x=[10,20,30,40,50,60,70,80,90]
a=[8,1,7,2,0,3,7,3,2]
```

```python
plt.scatter(x,a)
plt.show()
```

## Markers in the scatter plot

```python
plt.scatter(x,a, marker="*",c="orange",s=1000)
plt.show()
```

## adding two markers in the same plot

```python
z=[10,20,30,40,50,60,70,80,90]
x=[8,1,7,2,0,3,7,3,2]
y=[7,2,5,6,9,1,4,5,3]
```

```python
plt.scatter(z,x, marker="*",c="g",s=100)
plt.scatter(z,y, marker=".",c="r",s=200)
plt.show()
```
